# Gráficos de perfís de médias

```python
# Começamos carregando as bibliotecas necessárias para a nossa análise:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn

# Fixing random state for reproducibility
np.random.seed(2360873)
N = 100
```

```python
# Carregamos os dados de gorjetas presentes no Python como exemplo para fazer a nossa análise:
tips = sns.load_dataset("tips")

# Criamos uma nova feature chamada 'tip_pct', que é a porcentagem da gorjeta em relação ao total da conta do cliente:
tips['tip_pct'] = tips['tip'] / (tips['total_bill'] - tips['tip'])

# Mostramos o gráfico para visualização e verificação:
tips.head()
```

```python
ax = sns.pointplot(x = "sex", 
                   y = "tip_pct",
                   data = tips, 
                   dodge = True, 
                   errorbar = ('ci', 90)
                  )
```

## 🧾 Análise do gráfico de médias e intervalos de confiança (gorjetas por sexo)

### O gráfico apresenta a relação entre o sexo dos clientes (sex) e o percentual médio de gorjetas (tip_pct). Cada ponto indica a média de gorjeta para homens (Male) e mulheres (Female), e as barras verticais representam os intervalos de confiança de 90%, que mostram a incerteza associada a essas médias.

### Observa-se que as mulheres apresentam uma média de gorjeta ligeiramente maior que os homens, conforme o ponto correspondente no eixo y. No entanto, os intervalos de confiança se sobrepõem, o que indica que essa diferença não é estatisticamente significativa — ou seja, não é possível afirmar com segurança que as mulheres realmente recebem gorjetas proporcionalmente maiores que os homens.

### Além disso, o intervalo de confiança dos homens é mais amplo, sugerindo que há maior variabilidade nas gorjetas dadas a clientes do sexo masculino (alguns valores muito altos e outros muito baixos). Já o grupo feminino apresenta um intervalo um pouco mais estreito, indicando maior consistência nas gorjetas.

### De forma geral, o gráfico mostra que o sexo do cliente não exerce influência clara sobre o percentual das gorjetas, e que a diferença observada pode ser explicada pela variabilidade natural dos dados.

```python
ax = sns.pointplot(x="day", 
                   y="tip_pct", 
                   hue = 'sex',
                   data=tips, 
                   dodge=True, 
                   errorbar = ('ci', 95)
                  )

plt.legend(loc = 'lower center', 
           bbox_to_anchor = (0.5, -.30), 
           ncol = 3)
```

## 🧾 Análise do gráfico de médias e intervalos de confiança (gorjetas por dia e sexo)

### O gráfico apresenta a variação do percentual médio de gorjetas (tip_pct) ao longo dos dias da semana, comparando os grupos de homens (Male) e mulheres (Female). Cada ponto representa a média das gorjetas em determinado dia, enquanto as barras verticais correspondem aos intervalos de confiança de 95%, indicando a incerteza da estimativa da média.

### Observa-se que as médias de gorjetas são relativamente próximas entre homens e mulheres na maioria dos dias, com sobreposição dos intervalos de confiança, o que indica ausência de diferença estatisticamente significativa entre os grupos.

### Na sexta-feira, o grupo feminino apresenta média de gorjeta um pouco maior do que o masculino, porém com ampla variação (intervalo de confiança maior), sugerindo que as gorjetas das mulheres foram menos consistentes — houve tanto valores altos quanto baixos.

### Nos demais dias (quinta, sábado e domingo), as médias são bastante semelhantes e as barras de confiança se cruzam fortemente, reforçando a ideia de que o sexo do cliente não é um fator determinante no percentual das gorjetas.

### De modo geral, o comportamento das médias mostra que as variações entre os dias da semana influenciam mais o valor das gorjetas do que o sexo dos clientes. A sexta-feira e o domingo tendem a concentrar as maiores médias, enquanto quinta e sábado apresentam valores menores e mais estáveis.
