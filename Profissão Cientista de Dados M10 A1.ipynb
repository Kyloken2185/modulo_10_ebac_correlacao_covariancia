# Análise descritiva para modelos de regressão

Nesse módulo discutiremos algumas técnicas de análise descritiva específicas para situações em que temos uma variável de interesse (variável resposta) contínua. É uma boa prática colocar a variável de interesse no eixo **Y** (vertical) de um gráfico de dispersão, deixando a variável explicativa no eixo **X** (horizontal).

- Correlação e covariância
- Outliers
- *Ranking* e correlação de *Spearman*
- Matriz de correlação

```python
# Vamos buscar as bibliotecas necessárias para fazer a nossa análise:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Fixing random state for reproducibility
# Nesta linha de código colocamos uma "seed" que trava a sequência de números pseudoaleatórios garantindo a reprodutibilidade, ou seja, se 
# este código rodar em outra máquina, ou em outra execução, os números serão sempre os mesmos.  
np.random.seed(2360873)

# Aqui definimos o valor de uma variável que será usada mais adiante:
N = 100
```

```python
# Aqui determinamos N valores aleatórios para a variável x, sendo que N = 100: 
x = 0 + 8*np.random.rand(N)

# Em seguida, fazemos o mesmo para uma variável y
y = x + np.random.randn(N)*.5

# Colocamos estes valores em um DataFrame com suas respectivas colunas:
df1 = pd.DataFrame({'x':x, 'y':y})

# Aqui plotamos as variáveis no plano cartesiano, usando como parâmetros os valores da coluna 'x' no eixo-x, os valores da coluna 'y' no 
# eixo-y e como dados o DataFrame df1:
_ = sns.regplot(x = 'x', y = 'y', data = df1)

# Aqui aplicamos o método '.corr()' para descobrir a correlação entre as variáveis x e y:
df1.corr()
```

### No gráfico acima, podemos perceber que a correlação entre as variáveis x e y é próxima de 1. Isto significa que existe uma forte relação linear positiva entre elas. Quando uma cresce a outra tende a crescer também, seguindo um comportamento semelhante.

```python
x = 0 + 8*np.random.rand(N)
y = -1*x + np.random.randn(N)*.2
df2 = pd.DataFrame({'x':x, 'y':y})

_ = sns.regplot(x='x', y='y', data = df2)
df2.corr()
```
