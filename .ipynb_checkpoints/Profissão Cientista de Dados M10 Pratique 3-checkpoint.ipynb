### Primeiramente vamos carregar os pacotes necessários para a nossa análise:

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pandas.errors import SettingWithCopyWarning
import warnings
````

### Agora, vamos carregar o nosso Dataframe

````python
df = pd.read_csv("previsao_de_renda.csv")
````

### Para verificar se o DataFrame está corretamente carregado, vamos mostrá-lo na tela:

````python
df.head(5)
````

### Para começar a nossa análise, devemos saber de qual é o tipo de cada variável presente em nosso DataFrame:

````python
df.dtypes
````

### Vamos então construir um gráfico para cada variável qualitativa para investigar se houve alguma interação importante a ser investigada:

````python
# Vamos construir um DataFrame usando a coluna "qtd_filhos" com a primeira variável qualitativa "tempo_emprego" 
df1 = df[["posse_de_veiculo","tempo_emprego"]]

# Mostramos o DataFrame para visualização e verificação:
df1.head(5)
````

### Vamos mudar as variáveis "True" e "False" para "Possui" e "Não possui", usando o método "replace"

````python
# Apagando mensagens do tipo "warnings":
warnings.filterwarnings('ignore', category = SettingWithCopyWarning)

df1["posse_de_veiculo"] = df1["posse_de_veiculo"].replace({
    True: "Possui",
    False: "Não possui"
})

# Vamos imprimir o nosso DataFrame com o valor da variável "posse_de_veiculo" mudada:
df1.head()
````

### Vamos criar faixas de tempo para que possamos fazer o gráfico para a nossa análise. Para isto, devemos descobrir qual é o valor mínimo e o valor máximo da variável "tempo_emprego"

````python
tempo_min = df1["tempo_emprego"].min()
print("O tempo mínimo de emprego é:", tempo_min)

tempo_max = df1["tempo_emprego"].max()
print("O tempo máximo de emprego é:", tempo_max)
````

### Podemos perceber que há pessoas que ficaram meses na empresa e, ao contrário, há aquelas que ficaram até se aposentar. Então vamos dividir os valores do tempo de empresa em faixas: de 0 a 1 ano, de 1 ano a 3 anos, de 3 anos a 5 anos, de 5 anos a 10 anos, de 10 anos a 20 anos e mais de 20 anos. Estas faixas foram pensadas como fases de carreira:   

 Faixa | Significado        |
|-------|--------------------|
| 0-1   | recém contratado   |
| 1-3   | início de carreira |
| 3-5   | estabilização      |
| 5-10  | experiência        |
| 10-20 | senioridade        |
| 20+   | veterano           |

````python
# Apagando mensagens do tipo "warnings":
warnings.filterwarnings('ignore', category = SettingWithCopyWarning)
warnings.filterwarnings('ignore', category = FutureWarning)

# Criando faixas de tempo:
df1['faixa_tempo'] = pd.cut(
    df1['tempo_emprego'],
    bins = [0, 1, 3, 5, 10, 20],
    labels = ['0-1', '1-3', '3-5', '5-10', '10+']
)

# Agrupando a quantidade de funcionários por faixa de tempo de emprego:
df1_plot = df1.groupby(['faixa_tempo', 'posse_de_veiculo']).size().reset_index(name ='quantidade')

df1_plot.head()
````

### Para analisar, plotamos o gráfico de barras para verificar a distribuição da variável "posse_de_veiculo" por faixa de tempo:

````python
# Vamos definir o tamanho da figura em que será alocado o nosso gráfico:
plt.figure( figsize = (8, 6) )

# Abaixo, escrevemos o código para a construção de nosso gráfico de barras para estas duas variáveis:
ax = sns.barplot(
    x = 'faixa_tempo',
    y = 'quantidade',
    hue = 'posse_de_veiculo',
    data = df1_plot,
    palette = {
        "Não possui" : "red",
        "Possui" : "blue"
    }
)

ax.set_title("Relação entre a quantidade de funcionários propietários de veículos com o tempo de empresa", fontsize = 14, fontweight = "bold")
ax.set_xlabel("Tempo de empresa", fontweight = "bold")
ax.set_ylabel("Quantidade de funcionários", fontweight = "bold")
````

### Podemos perceber que tanto os funcionários recém-contratados, quanto aqueles que estão na empresa há muito tempo, não possuem veículos, assim como os estáveis na empresa situados na faixa de tempo de 3 a 10 anos, mostrando que o tempo de empresa não influencia nesta variável.
### Entretanto, o gráfico indica que quanto maior é o tempo de empresa, maior é a quantidade de funcionários proprietários de veículos. Somente caindo quando a faixa de tempo de empresa passa a ser de 10 anos ou mais.

### Vamos fazer a contagem da quantidade de "Bom pagador" e "Mau pagador" presentes em nosso DataFrame:

````python
print(f"A quantidade de bons e maus pagadores em nossa base de dados é:\n", df8["mau"].value_counts())
````

### A nossa base de dados indica que grande maioria dos funcionários são bons pagadores, mas gostaria de analisar em que faixa de tempo de empresa se situam os maus pagadores, pois temos a suspeita de que aqueles que têm menos tempo de empresa sejam a maioria dos maus pagadores.
### Para isto, vamos criar um filtro e selecionar somente aquelas linhas que contenham os "Maus pagadores":

````python
df8_mau = df8[df8["mau"] == "Mau pagador"].dropna().reset_index()

df8_mau.head(5)
````

### Dividindo o DataFrame em faixas de tempo, temos:

````python
# Apagando mensagens do tipo "warnings":
warnings.filterwarnings('ignore', category = SettingWithCopyWarning)
warnings.filterwarnings('ignore', category = FutureWarning)

# Criando faixas de tempo:
df8_mau['faixa_tempo'] = pd.cut(
    df8_mau['tempo_emprego'],
    bins = [0, 1, 3, 5, 10, 20],
    labels = ['0-1', '1-3', '3-5', '5-10', '10+']
)

# Agrupando a quantidade de funcionários por faixa de tempo de emprego:
df8_plot = df8_mau.groupby(['faixa_tempo', 'mau']).size().reset_index(name ='quantidade')

df8_plot.head()
````

### Nossas suspeitas parecem que estão sendo fundamentadas por esta primeira visão de nosso DataFrame separado por faixas de tempo de empresa. Isto poderá ser reforçado plotando os dados em nosso gráfico de barras. 

````python
# Vamos definir o tamanho da figura em que será alocado o nosso gráfico:
plt.figure( figsize = (8, 6) )

# Abaixo, escrevemos o código para a construção de nosso gráfico de barras para estas duas variáveis:
ax = sns.barplot(
        x = 'faixa_tempo',
        y = 'quantidade',
        data = df8_plot,
        color = "red"
)

ax.set_title("Relação entre a quantidade de funcionários maus pagadores com o tempo de empresa", fontsize = 14, fontweight = "bold")
ax.set_xlabel("Tempo de empresa", fontweight = "bold")
ax.set_ylabel("Quantidade de funcionários", fontweight = "bold")
````

## 2) Para cada variável qualitativa, construa um gráfico mostrando as médias de cada nível da variável no tempo e avalie a estabilidade. Houve alguma alteração importante? Você acha que se pode utilizar essas variáveis em uma modelagem?

## Vamos fazer gráficos que relacionam as variáveis qualitativas com a proporção de maus e bons pagadores, agrupando segundo estas variáveis e as faixas de tempo. Para isto, vamos transformar a variável "mau" em uma variável binária.

````python
# Selecionamos as variáveis necessárias para analisar a proporção média de inadimplência por posse de veículo e faixa de tempo:
df1_mau = df[["posse_de_veiculo","tempo_emprego","mau"]]

# Mudamos o tipo de variável da coluna "mau" para podermos calcular a proporção média de inadimplẽncia:
df1_mau["mau"] = df1_mau["mau"].replace({
    True : 1,
    False : 0
})

# Também mudamos os valores da variável "posse_de_veiculo":
df1_mau["posse_de_veiculo"] = df1_mau["posse_de_veiculo"].replace({
    True : "Posssui",
    False : "Não possui"
})

# Criando faixas de tempo:
df1_mau['faixa_tempo'] = pd.cut(
    df1_mau['tempo_emprego'],
    bins = [0, 1, 3, 5, 10, 20],
    labels = ['0-1', '1-3', '3-5', '5-10', '10+']
)

df1_mau.head()
````

````python
# Agrupamos a proporção média da variável "mau" por "faixa_tempo" e "posse_de_veiculo":
df1_mau_media = df1_mau.groupby(['faixa_tempo', 'posse_de_veiculo'])['mau'].mean().reset_index()

# Mostramos o DataFrame para visualização e verificação:
df1_mau_media.head()
````

````python
# Vamos definir o tamanho da figura em que será alocado o nosso gráfico:
plt.figure( figsize = (8, 6) )

# Abaixo, escrevemos o código para a construção de nosso gráfico de barras para estas duas variáveis:
ax = sns.barplot(
        x = 'faixa_tempo',
        y = 'mau',
        hue = 'posse_de_veiculo',
        data = df1_mau_media
)

ax.set_title("Média de inadimplência por posse de veículo ao longo do tempo", fontsize = 14, fontweight = "bold")
ax.set_xlabel("Faixa de tempo de empresa", fontweight = "bold")
ax.set_ylabel("Proporção de maus pagadores", fontweight = "bold")
ax.legend(title = 'Posse de veículo')
````

## Observa-se que funcionários que não possuem veículo apresentam maior proporção de inadimplência em todas as faixas de tempo de empresa, especialmente entre 1 e 3 anos.

## Apesar de existir certa variação temporal, principalmente no intervalo de 1 a 3 anos, o comportamento geral das categorias permanece relativamente consistente ao longo do tempo, indicando que a variável pode ser relevante para utilização em modelagem preditiva.

## Vamos fazer o mesmo método para analisar as variáveis "sexo", "tempo_emprego" e "mau"

````python
# Vamos construir um DataFrame usando a coluna "posse_de_imovel" com a primeira variável qualitativa "tempo_emprego" 
df3_mau = df[["sexo","tempo_emprego","mau"]]

# Mudamos o tipo de variável da coluna "mau" para podermos calcular a proporção média de inadimplẽncia:
df3_mau["mau"] = df3_mau["mau"].replace({
    True : 1,
    False : 0
})

# Mudamos o tipo de variável da coluna "sexo": "F" para "Feminino" e "M" para "Masculino":
df3_mau["sexo"] = df3_mau["sexo"].replace({
                        "F" : "Feminino",
                        "M" : "Masculino"})

# Criando faixas de tempo:
df3_mau['faixa_tempo'] = pd.cut(
   df3_mau['tempo_emprego'],
   bins = [0, 1, 3, 5, 10, 20],
   labels = ['0-1', '1-3', '3-5', '5-10', '10+']
)

df3_mau.head()
````

````python
# Agrupamos a proporção média da variável "mau" por "faixa_tempo" e "posse_de_veiculo":
df3_mau_media = df3_mau.groupby(['faixa_tempo', 'sexo'])['mau'].mean().reset_index()

# Mostramos o DataFrame para visualização e verificação:
df3_mau_media.head()
````

````python
# Vamos definir o tamanho da figura em que será alocado o nosso gráfico:
plt.figure( figsize = (8, 6) )

# Abaixo, escrevemos o código para a construção de nosso gráfico de barras para estas duas variáveis:
ax = sns.barplot(
        x = 'faixa_tempo',
        y = 'mau',
        hue = 'sexo',
        data = df3_mau_media,
        palette = {
            "Feminino" : "pink",
            "Masculino" : "blue"
        }
)

ax.set_title("Média de inadimplência por sexo ao longo do tempo", fontsize = 14, fontweight = "bold")
ax.set_xlabel("Faixa de tempo de empresa", fontweight = "bold")
ax.set_ylabel("Proporção de maus pagadores", fontweight = "bold")
ax.legend(title = 'sexo')
````

````python
## Observa-se que, apesar da inversão temporária na faixa de 1 a 3 anos, o comportamento das categorias permanece relativamente consistente ao longo do tempo.

## Em grande parte das faixas analisadas, funcionários do sexo masculino apresentam maior proporção de inadimplência em comparação aos funcionários do sexo feminino.

## Dessa forma, a variável “sexo” demonstra relativa estabilidade temporal e pode apresentar potencial para utilização em modelagem preditiva.
````
